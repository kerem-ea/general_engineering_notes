## Propositions
A proposition is a statement with a truth value (True or False, 1 or 0). Examples: (3 > 2),  (1 = 5) etc. \
A propositional function (a predicate) is for example: $x = 2$ or $x^2 = y-1$. 

They don't have a truth value (aren't a proposition) until you input values. 

## Truth values
A padlock with 3 dials: \
$P : x = 3$ \
$Q : y = 4$ \
$R : z = 7$ 

| P | Q | R |
| --| --| --|

For each proposition n the possible combinations of the propositions are equal to: $2^n$



## Logical operations
If we want P, Q and R to be true at the same time for the lock to be open, we use what is called a conjunction (AND or $\land$). \
$P \land Q \land R$

If we only need either P or Q to be true, we can use what is called a disjunction (OR or $\lor$). \
$P \lor Q$ 

There is also a negation, which is written as $\neg P$. It is defined as NOT P.

## Logical equivalence
$(Q \land \neg P) \lor P$ is logically equivalent to $Q \lor P$. 

| $P$ | $Q$ | $\neg P$ | $Q \lor \neg P$| $(Q \lor \neg P) \lor P $| $Q \lor P$ |
| --| --| --| --| --| --|

$P \land P$ is logically equivalent to $P$. \
$P \lor P$ is logically equivalent to $P$. \
$P \land Q$ is logically equivalent to $Q \land P$. 
$P \lor Q$ is logically equivalent to $Q \lor P$. 

### De Morgan's Law
If you negate an $\land$ or an $\lor$ you can negate whatever is inside the parenthesis and turn the $\lor$ into an $\land$ (and reverse). 

Multiplication using binary numbers (0 and 1) corresponds to $\land$. 

## Logical consequence
When an implication is a tautology, it is a logical consequence. 

## Implications
An implication is when a proposition has an effect on another propositions truth values. \
It is also called "if". \
For example: 
$P \to Q$, \
means that if $P$ happens, $Q$ must follow. If $P$ happens and $Q$ doesn't follow, it is false. \
If $P$ doesn't happen and $Q$ happens, the promise isn't broken so it is true. \
If both are true or both are false the implication is true. \
A biimplication is is when $P \to Q$ and $Q \to P$ both are true.


In other words, the two operations MUST have the same truth values. \
This is the denoted by the symbol: $\iff$. \
If all the same truth values are $\top$, it is a tautology.

## Tautologies
A tautology is when all truth values are true. \
$\neg P \lor P$ is always true, since they are always opposite of each other.
This is logically equivalent to $\top$.

$P \land \top$ is logically equivalent to $P$.

## Contradictions
A contradiction is when all truth values are false. \
$\neg P \land P$ is always false, since they are always opposite of each other. \
This is logically equivalent to $\bot$. (F) \
$P \land \bot$  is logically equivalent to $\bot$ (F)
