#### Question 5
A cylindrical bottle with a height of 21.5 cm is filled with an olive oil of density 0.953 g/mL. 1360 g of oil is needed to fill the bottle. What is the diameter of the bottle?
    1. 8.65 cm
    2. 9.20 cm
    3. 9.95 cm
    4. 8.20 cm

    Answer: 2.
            A cylinder has the volume equal to: 
                V = r^2 * Pi * height
            We can find the volume by using the information regarding the olive oil density and amount needed to fill the bottle. 
                density = 0.953 g/mL
                mass = 1360g
                V = mass/density
    
            With the volume found we can isolate the radius in the volume formula:
                r = sqrt(V/height*Pi)
            We can then multiply the radius by 2 to find the diameter:
                d = 2r






```python
import math  
density = 0.953 
mass = 1360 
height = 21.5  

volume = mass / density 
r = (volume / (height * math.pi))**0.5 
d = 2 * r 

print(d)
```

#### Question 6
The annual production of Mg metal in USA is 80000 tons per year (US tons).  Mg is obtained mostly from seawater, which has a density of 1.03 g/mL and contains 1300 mg of Mg per kg of seawater. Calculate the volume of the seawater needed to produce the Mg equal to the annual production of USA.
    1. 5.4 * 10^12 L
    2. 5.4 * 10^9 L
    3. 5.4 * 10^10 L
    4. 5.4 * 10^11 L

Answer: 3. We know the annual production of Mg metal in the USA (80000 US tons). We want to know the amount in kg, and the conversion from US tons to kg is: 
    1 US ton = 907.185kg

The density of seawater is:    
    density = 1.03 g/mL

One kg of seawater contains 1300mg (1.3g) of Mg. We need to know how much seawater we need to get 80000 US tons of Mg.


```python
tons = 80000
ton_to_kg = 907.185
mg_per_kg = 1.3
density = 1.03

mass_mg = tons * ton_to_kg
mass_seawater = mass_mg / (mg_per_kg / 1000)
volume = mass_seawater / density

print(volume)
```

#### Question 7
Density of aluminium is 2.699 g/cc. A sheet of aluminium with a total area of 1.0 ft^2 has a mass of 3.636 g. Calculate the thickness of the aluminium sheet.
    1. 4.45 * 10^-3 mm
    2. 1.45 * 10^-3 mm
    3. 4.45 * 10^-2 mm
    4. 1.45 * 10^-2 mm

Answer: The density of aluminium is:
    d = 2.69/cc
    1 ft^2 of aluminium has a mass of 3.636g.
    1 ft^2 = 0.093 m^2
    1 m^2 = 10000 cm^2
    volume = mass / density
    thickness = volume / area
    
    



```python
density = 2.699       # g/cm^3
mass = 3.636          # g

# Convert 1 ft^2 to m^2
area_m2 = 0.093

# Convert m^2 to cm^2
area_cm2 = area_m2 * 10000

# Find volume
volume_cm3 = mass / density

# Find thickness in cm
thickness_cm = volume_cm3 / area_cm2

# Convert cm to mm
thickness_mm = thickness_cm * 10

print("Area =", area_cm2, "cm^2")
print("Volume =", volume_cm3, "cm^3")
print("Thickness =", thickness_mm, "mm")
print("Answer = {:.2e} mm".format(thickness_mm))
```

#### Question 8
The densities of Cu and Sn are 8.94 g/cm3 and 7.31 g/cm3 respectively. An homogeneous Cu-Sn alloy has 79.42% Cu and 20.58% Sn. What is the weight (mass) of a Cu-Sn alloy cylinder of radius 64.4 mm and a length of 443.7 mm.
    1. 49.7kg
    2. 59.7kg
    3. 55.7kg
    4. 53.7kg

Answer:
d_cu = 8.94 g/cm^3
d_sn = 7.31 g/cm^3
Cu-Sn alloy is 0.7942 Cu and 0.2058 Sn.
r = 64.4 mm = 6.44 cm
length = 443.7mm = 44.37 cm

V = r^2 * length * Pi
V = (6.44)^2 * 44.37 * Pi
V = 5776.3 cm^3

m_cu = 0.7942 * V * d_cu
m_sn = 0.2058 * v * d_sn

m = m_cu + m_sn

```python
import math

# Given data
d_cu = 8.94          # Density of Cu, g/cm^3
d_sn = 7.31          # Density of Sn, g/cm^3

v_cu = 0.7942        # Volume fraction of Cu
v_sn = 0.2058        # Volume fraction of Sn

r_mm = 64.4          # Radius, mm
length_mm = 443.7    # Length, mm


# Convert dimensions from mm to cm
r_cm = r_mm / 10
length_cm = length_mm / 10


# Calculate volume of cylinder
V = math.pi * r_cm**2 * length_cm


# Calculate mass of each component
mass_cu = v_cu * V * d_cu
mass_sn = v_sn * V * d_sn


# Calculate total mass
total_mass_g = mass_cu + mass_sn
total_mass_kg = total_mass_g / 1000


# Display results
print(f"{'Cu volume percentage:':<25} {v_cu * 100:>10.2f}%")
print(f"{'Sn volume percentage:':<25} {v_sn * 100:>10.2f}%")
print(f"{'Radius:':<25} {r_cm:>10.2f} cm")
print(f"{'Length:':<25} {length_cm:>10.2f} cm")
print(f"{'Cylinder volume:':<25} {V:>10.2f} cm^3")
print(f"{'Mass of Cu:':<25} {mass_cu:>10.2f} g")
print(f"{'Mass of Sn:':<25} {mass_sn:>10.2f} g")
print(f"{'Total mass:':<25} {total_mass_g:>10.2f} g")
print(f"{'Total mass:':<25} {total_mass_kg:>10.2f} kg")

```
