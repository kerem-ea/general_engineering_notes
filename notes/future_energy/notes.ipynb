### Energy
Energy is constant (can't be created or destroyed, only altered). \
Energy is measured in joules (J). \
There are many forms of energy: \
    1. Kinetic energy \
    2. Thermal energy \
    3. Potential energy \
    4. Solar energy \
    5. etc. etc. \
There are only four fundamental forms of energy: \
1. Kinetic energy
2. Gravitational energy
3. Electrical energy
4. Nuclear energy

### Work
Work (W) is defined as force times displacement. \
It is also defined as ordered movement against an opposing force, for example moving against gravity dragging a box down. \
If the force is constant: \
$W = F \times x$ \
If the force is not constant it can be written as: \
$W = \int\limits_a^b F dx$ 

### Force
A force is defined as: \
$F = m \times a$ \
A force may transfer energy if displacement happens in the direction of the force. \
If there is no displacement, the force didn't affect any energy. 

### Power
Power is the flow of energy, energy over time. 
$P = \frac{E}{t}$ \
where E is energy and t is time. 

### Newton's laws
1. "An object at rest stays at rest and an object in motion stays in motion with the same speed and direction until acted upon by an external force."
2. "The acceleration of an object is directly proportional to the net force acting on it and inversely proportional to its mass ($F = m \times a$)."
3. "For every action force, there is an equal and opposite reaction force".

### Gravitational energy
Mass attraction: $F = G\frac{m1 \times m2 }{r^2}$ \
m1 and m2 are the masses of the bodies, \
r is the distance, \ 
G is the gravitational constant, \
g is the gravitational force (approx. 9.82) \
$E_{pot} = m \times g \times h$

### Electrical energy
$F = k_e \frac{q1 \times q2 }{r^2}$ \
$k_e$ is the coulomb constant \
q is the charges \
r is the distance \

### Nuclear energy
Mass and energy are equivalent (same thing):
$E = m \times c^2$ \
where m is the mass and c is the speed of light.

### Three fundamental energy regimes
1. Macroscopic scale: wind, heat, hydropower tides etc. 
2. Molecular scale: thermal motion, chemical reactions etc.
3. Subatomic scale: particle radiation, electromagnetic radiation, nuclear reactions etc.

### Capacity factor (CF)
CF is the quantification of the utilization of a device. \
$CF = \int \frac{Actualpower}{Rated power}dT$ \
$CF = \frac{Actual energy produced}{Rated power \times time}$

### Higher and lower heating value
You always liberate energy when you burn a fuel, otherwise it's not a fuel. \
If you combust a fuel and you get vapor, then you can collect that steam and condense it and get more energy out of it. \
If you get liquid water: more energy, if you get steam water: less energy.  

### Ordered and disordered energy (work & heat)
Heat is disordered thermal energy tranferred to/from the system. Atoms and molecules are accelerated randomly. \
Ordered mechanical (electrical) energy transferred to/from the system. Atoms and molecules are accelerated ordered. 

### Pressure
Pressure is defined as force per area: \
$Pressure = \frac{N}{m^2}$ \
In a powerplant you can expand gas which creates pressure on for example a piston which would turn heat (unuseable kinetic energy) into useful kinetic energy. \
$W = p \times \Delta V$ (should be in differential form)

### Ideal gas law
$pV = nRT$ \
where p is the pressure, V is the volume, n is the amount of gas (mol), R is the gas constant and T is the temperature. \
Assumes no interaction between molecules apart from elastic collisions and that the volume of each molecule is negligible.
